```python
# matrix operations

import numpy as np

def sum_matrix(left_matrix: np.ndarray, right_matrix: np.ndarray) -> None:
    result = left_matrix + right_matrix
    print(result)  

def substract_matrix(left_matrix: np.ndarray, right_matrix: np.ndarray) -> None:
    result = left_matrix - right_matrix
    print(result)  

def multiply_two_matrices(left_matrix: np.ndarray, right_matrix: np.ndarray) -> np.ndarray:
    result = np.dot(left_matrix, right_matrix)
    print(result)
    return result

def multiply_three_matrices(first_matrix: np.ndarray, second_matrix: np.ndarray, third_matrix: np.ndarray) -> None:
    first_multiplication_result = multiply_two_matrices(first_matrix, second_matrix)
    result = np.dot(first_multiplication_result, third_matrix)
    print(result)  
```

```python
a = np.array([[1, 2, 4],
              [7, 5, -2],
              [-2, 1, 3]
             ])
b = np.array([[-1, 0, 3],
              [2, -6, 1],
              [-7, 1, 5]
             ])
c = np.array([[-2, 3],
              [4, 0],
              [7, -1]
             ])

print("A+B\n")
sum_matrix(a, b)
print("B-A\n")
substract_matrix(b, a)
print("A*C\n")
multiply_two_matrices(a, c)
print("A*B*C\n")
multiply_three_matrices(a, b, c)
```
