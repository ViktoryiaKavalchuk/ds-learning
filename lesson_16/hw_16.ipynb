```python
from sklearn.model_selection import train_test_split

#Fields for model with high correlation:
data_for_training = data.copy(deep=True)
data_for_training = data_for_training.drop(['id','date'], axis='columns')

x_columns = [x_column for x_column in data_for_training.columns if x_column != "price"]
#x_columns = ['bedrooms', 'bathrooms', 'sqft_living', 'floors', 'grade', 'sqft_above', 'yr_built', 'long', 'sqft_living15', 'sqft_lot15']

x, y = data[x_columns], data['price']
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.15, random_state=42)
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import f1_score

reg = LinearRegression().fit(x_train, y_train)

print(f"Score - {reg.score(x_test, y_test)}")
print(f"Coefficient - {reg.coef_}")
print(f"Intercept - {reg.intercept_}")

y_predicted = reg.predict(x_test)

print(f"MAE - {mean_absolute_error(y_test, y_predicted)}")
print(f"MSE - {mean_squared_error(y_test, y_predicted)}")
```
